```python
import os
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt
```

```python
n = 30

dossier_img = Path.home() / f"UNET_DATA/images"
dossier_gt = Path.home() / f"UNET_DATA/gt"

nom_img = os.listdir(dossier_img)[n]
nom_gt = os.listdir(dossier_gt)[n]

chemin_img = os.path.join(dossier_img, nom_img)
chemin_gt = os.path.join(dossier_gt, nom_gt)

img = Image.open(chemin_img)
gt = Image.open(chemin_gt)

fig, axes = plt.subplots(1, 2)
axes[0].imshow(gt)
axes[1].imshow(img)
```
